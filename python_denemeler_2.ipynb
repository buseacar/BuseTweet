```python
liste=[]
tek_sayilar_listesi=[]
sayac1=0
sayac2=0

for i in range(1,10+1):
  sayi=input("bir sayı giriniz")

  if sayi.strip() == "":
        print("yanlış tuşlama")
        continue
  sayi=int(sayi)
  liste.append(sayi)
  sayac1 += 1
  print(sayac1)

for sayi in liste:
  if sayi%2==1:
    tek_sayilar_listesi.append(sayi)
    sayac2+=1
print("tek sayılar:", tek_sayilar_listesi)
```

```python
liste = []  #bu da aynısının yanlış tuşlamayı sayaca katmadığı wheel versiyonu
sayac = 0

while sayac < 10:
    sayi = input("bir sayı giriniz: ")

    if sayi.strip() == "":
        print("yanlış tuşlama, tekrar dene")
        continue

    try:
        sayi = int(sayi)
    except ValueError:
        print("sayı girmen lazım")
        continue

    liste.append(sayi)
    sayac += 1
    print("girilen doğru sayı:", sayac)
```
